**Ziel / Objective:**

- 🇬🇧 Performing Transfer Learning (Fine-tuning) with YOLOv11 nano directly on the NVIDIA Jetson GPU. Monitoring Box Loss, Class Loss, and mAP50.

### 1. Training starten via Ultralytics Python API / CLI

```python
from ultralytics import YOLO
import os

# Modell laden (YOLOv11n pretrained)
model = YOLO('yolo11n.pt')

# Transfer-Learning Training (z.B. 10 Epochen für Praktikums-Durchlauf)
results = model.train(
    data='/workspace/student_data/dataset/data.yaml',
    epochs=10,
    imgsz=640,
    batch=8,
    project='/workspace/student_data/runs',
    name='hfu_yolo11_model',
    exist_ok=True
)

print("[DE] Training erfolgreich beendet! Trainierte Gewichte gespeichert.")
print("[EN] Training successfully finished! Trained weights saved.")
```

### 2. Trainingskurven & Verluste analysieren

```python
import cv2
import matplotlib.pyplot as plt

results_png = '/workspace/student_data/runs/hfu_yolo11_model/results.png'
if os.path.exists(results_png):
    img = cv2.imread(results_png)
    plt.figure(figsize=(12, 8))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("HFU YOLOv11 Training Curves (mAP & Loss)")
    plt.axis('off')
    plt.show()
```
