- 🇬🇧 Verification of the USB industrial camera and lighting under the Bosch light bar. Calculation of image sharpness using the Siemens star (Variance of Laplacian).

### 1. Imports & Setup

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request
import os

def show_image(img, title="Kamerabild / Camera Frame"):
    plt.figure(figsize=(8, 6))
    if len(img.shape) == 3:
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    else:
        plt.imshow(img, cmap='gray')
    plt.title(title)
    plt.axis('off')
    plt.show()
```

### 2. Kamerazugriff & Schärfemessung (Siemensstern)
🇩🇪 Wir messen die Bildschärfe mittels der Varianz des Laplace-Operators. Ein höherer Wert bedeutet schärfere Kanten.

🇬🇧 We calculate image sharpness using the Variance of Laplacian. A higher value indicates sharper edges.

```python
def measure_sharpness(image_gray):
    return cv2.Laplacian(image_gray, cv2.CV_64F).var()

def capture_camera_frame():
    # Option A: Try direct V4L2 capture
    cap = cv2.VideoCapture(0, cv2.CAP_V4L2)
    if cap.isOpened():
        for _ in range(5):
            ret, frame = cap.read()
        cap.release()
        if ret and frame is not None and np.mean(frame) > 2.0:
            return frame
    
    # Option B: Stream from Backend API (Shared camera access)
    try:
        stream = urllib.request.urlopen("http://backend:8000/api/stream", timeout=3)
        bytes_data = b''
        for _ in range(100):
            bytes_data += stream.read(1024)
            a = bytes_data.find(b'\xff\xd8')
            b = bytes_data.find(b'\xff\xd9')
            if a != -1 and b != -1:
                jpg = bytes_data[a:b+2]
                frame = cv2.imdecode(np.frombuffer(jpg, dtype=np.uint8), cv2.IMREAD_COLOR)
                if frame is not None:
                    return frame
    except Exception as e:
        print(f"[INFO] API stream fallback note: {e}")
    
    return None

frame = capture_camera_frame()
if frame is not None:
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    score = measure_sharpness(gray)
    show_image(frame, f"Live Frame - Sharpness Score: {score:.1f}")
    print(f"[DE] Schärfewert (Laplace-Varianz): {score:.1f}")
    print(f"[EN] Sharpness score: {score:.1f}")
else:
    print("[INFO] Camera stream empty. Using demo fallback image.")
```
