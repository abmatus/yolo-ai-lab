### 1. Generation der `data.yaml` Konfiguration

```python
import yaml
import os

dataset_config = {
    'path': '/workspace/student_data/dataset',
    'train': 'images/train',
    'val': 'images/val',
    'names': {
        0: 'scalpel_skalpell',
        1: 'forceps_pinzette',
        2: 'clamp_klemme'
    }
}

yaml_path = '/workspace/student_data/dataset/data.yaml'
with open(yaml_path, 'w') as f:
    yaml.dump(dataset_config, f, default_flow_style=False)

print(f"[DE] `data.yaml` erfolgreich erstellt: {yaml_path}")
print(f"[EN] `data.yaml` successfully generated: {yaml_path}")
```
