### 1. Laden des Ultralytics YOLOv11 Modells

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# Laden des vortrainierten YOLOv11 nano Modells
model = YOLO('yolo11n.pt')
print("[DE] YOLOv11 Modell erfolgreich geladen!")
print("[EN] YOLOv11 pre-trained model successfully loaded!")
```

### 2. Inferenz & Schwellenwert-Anpassung (Confidence Threshold)
🇩🇪 Erproben Sie, wie sich das Verändern von `conf=0.25` bis `conf=0.75` auf die Erkennungsrate auswirkt.

🇬🇧 Observe how changing `conf=0.25` to `conf=0.75` affects detection rates.

```python
# Inferenz auf Testbild / Live-Kamerabild
results = model.predict(source='https://ultralytics.com/images/bus.jpg', conf=0.5)

# Ergebnisse visualisieren
res_plotted = results[0].plot()
plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(res_plotted, cv2.COLOR_BGR2RGB))
plt.title("YOLOv11 Pre-trained Detection (COCO)")
plt.axis('off')
plt.show()
```
